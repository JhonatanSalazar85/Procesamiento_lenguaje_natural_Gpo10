*   **Pregunta 1.** 

Busca y elimina todos los saltos de línea '\n' que se encuentran al final de cada comentario.

```python
# Pregunta 1: Eliminar saltos de línea '\n' al final de cada comentario.

# Opción recomendada usando Regex (re.sub)
docs = [re.sub(r'\n$', '', comentario) for comentario in docs]

# Alternativa simple usando el método .strip() de Python
# docs = [comentario.strip() for comentario in docs]

```

*   **Pregunta 2.**  

Busca e imprime todas las palabras que terminan con dos o más signos de admiración seguidos, por ejemplo "!!!". 

Debes imprimir tanto la palabra como la totalidad de signos de admiración que le siguen. 

Indica cuántos resultados obtuviste.



```python
# Pregunta 2: Buscar palabras que terminan con dos o más signos de admiración (!!)
import re

pattern_exclamation = r'\w+!{2,}'

hallazgos_admiracion = []

for comentario in docs:
    # re.findall busca todas las ocurrencias en cada línea y las devuelve en una lista
    coincidencias = re.findall(pattern_exclamation, comentario)
    hallazgos_admiracion.extend(coincidencias)

# Imprimimos los resultados obtenidos
print("Palabras y signos encontrados:")
print(hallazgos_admiracion)

```

```python
# total de resultadosT
print(f"\nTotal de resultados obtenidos: {len(hallazgos_admiracion)}")
```

*   **Pregunta 3.**  

Busca e imprime todas las palabras que están escritas totalmente en mayúsculas. Cada coincidencia debe ser una sola palabra.

Indica cuántas palabras encontraste.



```python
# Pregunta 3 - Parte A: Buscar palabras totalmente en mayúsculas
import re

pattern_all_caps = r'\b[A-Z]+\b'

palabras_mayusculas = []

for comentario in docs:
    # Extraemos las coincidencias de cada comentario
    coincidencias = re.findall(pattern_all_caps, comentario)
    # Usamos extend para mantener una lista plana de strings [cite: 601]
    palabras_mayusculas.extend(coincidencias)

# Imprimimos los resultados obtenidos
print("Palabras encontradas en mayúsculas:")
print(palabras_mayusculas)
```

```python
# Pregunta 3 - Parte B: Indicación de la cantidad de resultados
total_encontradas = len(palabras_mayusculas)

print(f"Se encontraron un total de {total_encontradas} palabras totalmente en mayúsculas.")
```

*   **Pregunta 4.**  

Busca e imprime los comentarios en donde todos los caracteres alfabéticos (letras) están en mayúsculas. 

Cada coincidencia encontrada debe ser todo el comentario/enunciado.

Indica cuántos resultados obtuviste.


```python
# Identificación de comentarios en mayúsculas
import re

pattern_full_caps = r'^[^a-z]*$'

# Filtramos la lista 'docs' usando re.match para validar el enunciado completo
comentarios_solo_mayusculas = [c for c in docs if re.match(pattern_full_caps, c)]

# Imprimir los comentarios encontrados
print("Comentarios detectados totalmente en MAYÚSCULAS:")
for i, comentario in enumerate(comentarios_solo_mayusculas):
    print(f"{i+1}.- {comentario}")
```

```python
# Indicar la cantidad de resultados
total_comentarios_caps = len(comentarios_solo_mayusculas)
print(f"\nTotal de comentarios que cumplen la condición: {total_comentarios_caps}")
```

*   **Pregunta 5.**  

Busca e imprime todas las palabras que tengan una vocal acentuada, del tipo á, é, í, ó, ú. 

Indica cuántos resultados obtuviste.

```python
# Buscar palabras con vocales acentuadas (á, é, í, ó, ú)
import re

pattern_accents = r'\b\w*[áéíóúÁÉÍÓÚ]\w*\b'

palabras_acentuadas = []

for comentario in docs:
    # re.findall extrae todas las palabras que cumplen el patrón en el comentario actual [cite: 585]
    coincidencias = re.findall(pattern_accents, comentario)
    # Consolidamos en una lista plana [cite: 601]
    palabras_acentuadas.extend(coincidencias)

# Imprimir los hallazgos
print("Palabras con vocales acentuadas encontradas:")
print(palabras_acentuadas)

```

```python
# Ccantidad de resultados obtenidos
print(f"\nTotal de resultados obtenidos: {len(palabras_acentuadas)}")
```

*   **Pregunta 6.**  

Busca e imprime todas las cantidades numéricas monetarias, enteras o con decimales, que inician con el símbolo $\$$. 

Indica cuántos resultados obtuviste.

```python
# Buscar cantidades monetarias que inician con $
import re


pattern_monetary = r'\$\d+(?:\.\d+)?'

hallazgos_monetarios = []

for comentario in docs:
   
    coincidencias = re.findall(pattern_monetary, comentario)
    
    hallazgos_monetarios.extend(coincidencias)

# Imprimimos las cantidades encontradas
print("Cantidades monetarias detectadas:")
print(hallazgos_monetarios)

```

```python
# Cantidad total de resultados
print(f"\nTotal de resultados obtenidos: {len(hallazgos_monetarios)}")
```

*   **Pregunta 7.**

Indica cuántos resultados obtuviste.

*   **Pregunta 8.**  

Busca e imprime todas las palabras, variantes de "so" y "good", que tengan dos o más "o" en "so" y 3 o más "o" en good. 

Indica cuántas encontraste.

*   **Pregunta 11.**  

Busca e imprime la secuencia de dos o más palabras que están separadas por un guion, "-", sin que tengan espacios en blanco entre ellas.

Por ejemplo "Go-Kart" sería válido, pero "Go  -Kart" o "Go  -  Kart" no lo serían. 

Indica la cantidad de resultados obtenidos.

```python
import re

# Definir el patrón para localizar secuencias de letras unidas por guiones.

patron_guion = r'\b[a-zA-Z]+-[a-zA-Z]+(?:-[a-zA-Z]+)*\b'

hallazgos_guion = []

for comentario in docs:
    # Extraigo todas las secuencias que cumplen con mi patrón de guiones.
    coincidencias = re.findall(patron_guion, comentario)
    # Uso el método extend para mantener mi lista final plana y facilitar el conteo.
    hallazgos_guion.extend(coincidencias)

# Imprimo los resultados que logré identificar en el texto.
print("Secuencias de palabras con guion encontradas:")
print(hallazgos_guion)
```

*   **Pregunta 12.**  

Busca e imprime todas las palabras que terminan en "ing" o "ed". 

Indica la cantidad de palabras que encontraste de cada una.

```python
import re

# Defino los patrones para buscar palabras con las terminaciones solicitadas.
patron_ing = r'\b\w+ing\b'
patron_ed = r'\b\w+ed\b'

# Inicializo mis listas para almacenar los hallazgos de forma independiente.
hallazgos_ing = []
hallazgos_ed = []

# Recorro cada comentario en mi corpus 'docs'.
for comentario in docs:
    # Extraigo las palabras que terminan en 'ing', ignorando mayúsculas/minúsculas para mayor robustez.
    coincidencias_ing = re.findall(patron_ing, comentario, flags=re.IGNORECASE)
    # Extraigo las palabras que terminan en 'ed'.
    coincidencias_ed = re.findall(patron_ed, comentario, flags=re.IGNORECASE)
    
    # Agrego los resultados a mis listas globales usando extend() para mantener una estructura plana.
    hallazgos_ing.extend(coincidencias_ing)
    hallazgos_ed.extend(coincidencias_ed)

# Imprimo las palabras encontradas con la terminación 'ing'.
print("Palabras terminadas en 'ing':")
print(hallazgos_ing)

# Imprimo las palabras encontradas con la terminación 'ed'.
print("\nPalabras terminadas en 'ed':")
print(hallazgos_ed)

```

```python
# Finalmente, muestro la cantidad total que encontré para cada categoría.
print(f"\nCantidad de palabras terminadas en 'ing': {len(hallazgos_ing)}")
print(f"Cantidad de palabras terminadas en 'ed': {len(hallazgos_ed)}")
```

#**Parte 3. Proceso de limpieza.**

*   **Pregunta 13.**  

Ahora realiza un proceso de limpieza del corpus que incluya los siguientes procesos:

*   Solo se deben considerar caracteres alfabéticos. Es decir, se eliminan todos los signos de puntuación y caracteres especiales.
*   Todos los caracteres alfabéticos se transforman a minúsculas.
*   Se deben eliminar todos los espacios en blanco adicionales que se puedan encontrar en cada comentario.

Al finalizar dicho proceso de limpieza, imprime el resultado de los primeros 10 comentarios resultantes. 
   




```python
import re

# 1. Inicio el proceso de limpieza del corpus.
# Utilizo una lista nueva para no alterar mis datos originales (docs).
docs_clean = []

# Recorro cada comentario en mi lista original.
for comentario in docs:
    # Primero, utilizo re.sub para eliminar todo lo que NO sea un carácter alfabético o espacio.
    # Esto cumple con eliminar signos de puntuación y caracteres especiales[cite: 588, 694].
    solo_letras = re.sub(r'[^a-zA-Z\s]', '', comentario)
    
    # Segundo, transformo todo el texto resultante a minúsculas.
    en_minusculas = solo_letras.lower()
    
    # Tercero, elimino los espacios en blanco adicionales.
    # split() separa por cualquier cantidad de espacios y join() los une con un solo espacio[cite: 703].
    limpio = " ".join(en_minusculas.split())
    
    # Agrego el resultado procesado a mi nueva lista.
    docs_clean.append(limpio)

```

```python
# 2. Imprimo los resultados de los primeros 10 comentarios para verificar la limpieza.
print("Visualizo los primeros 10 comentarios tras la limpieza técnica:")
for i, resultado in enumerate(docs_clean[:10]):
    print(f"{i+1}.- {resultado}")
```

*   **Pregunta 14.**  

Con el resultado de la limpieza obtenido en la pregunta anterior, realiza ahora un proceso de tokenización por palabras del corpus. 

Es decir, al final de este proceso de tokenización, debes tener como resultado una lista de listas, donde cada comentario estará tokenizado por palabras.

Al terminar calcula el total de tokens obtenido en todo el corpus. 

```python

corpus_tokenizado = [comentario.split() for comentario in docs_clean]

total_tokens = sum(len(comentario) for comentario in corpus_tokenizado)

print(f"Total de tokens en el corpus: {total_tokens}")

```
